## 1. 情感分析




## 1.1 基本用法

```python
from transformers import pipeline
```

```python
# 方法1：明确指定模型名称； 如果不指定模型的名字，会下载一个默认模型； distilbert/distilbert-base-uncased-finetuned-sst-2-english
classifier = pipeline(
    task='sentiment-analysis',
    model='distilbert/distilbert-base-uncased-finetuned-sst-2-english'
)
result = classifier("I love using Hugging Face libraries! ")
print(result)
```

```python
chi = classifier("理想很丰满")
print(chi)
```

## 1.2 使用中文模型

```python
from transformers import pipeline

# Load the classification pipeline with the specified model
pipe = pipeline("text-classification", model="tabularisai/multilingual-sentiment-analysis")

# Classify a new sentence
sentence = "理想很丰满"
result = pipe(sentence)

# Print the result
print(result)
```

```python
sentence = "王爷当上了皇帝"
result = pipe(sentence)

# Print the result
print(result)
```

## 1.3 设置Huggingface镜像站

```python
#设置镜像
import os
os.environ['HF_ENDPOINT'] = 'https://hf-mirror.com'
```

## 1.4 模型缓存位置

> 下载的模型会保存到本地位置

1. 默认位置：
    - linux/macOs ~/.cache/huggingface/hub
    - Windows: C:\Users<用户名>.cache\huggingface\hub (Windows)
2. 修改位置
    - `HF_HOME`：改动全局HF缓存，【模型】和【数据集】的位置都修改
    - `TRANSFORMERS_CACHE`: 只修改【模型】保存的位置

## 1.5 多文本处理

```python
from transformers import pipeline
# 设置镜像
import os
os.environ['HF_ENDPOINT'] = 'https://hf-mirror.com'

classifier = pipeline(
    'sentiment-analysis',
    model='distilbert/distilbert-base-uncased-finetuned-sst-2-english'
)

# 批量处理
texts = [
    "I love using Hugging Face libraries!",
    "This is terrible.",
    "Not bad, could be better."
]

results = classifier(texts)
for text, result in zip(texts, results):
    print(f"Text: {text}")
    print(f"Result: {result}")
    print("-" * 50)
```

## 1.6 返回所有标签的结果

```python
from transformers import pipeline

classifier = pipeline(
    'sentiment-analysis',
    model='distilbert/distilbert-base-uncased-finetuned-sst-2-english',
    top_k=None, # 返回前k个结果，None表示返回所有标签的概率
)

result = classifier("I love using Hugging Face libraries!")
print(result)
```

### 1.6.1 下载单个文件

```python
from huggingface_hub import hf_hub_download

model_path = hf_hub_download(repo_id='distilbert/distilbert-base-uncased-finetuned-sst-2-english',
                             filename="config.json")
print(model_path)
```

### 1.6.2 下载整个模型

```python
from huggingface_hub import snapshot_download
model_path = snapshot_download(repo_id='distilbert/distilbert-base-uncased-finetuned-sst-2-english')
print(model_path)
```

```python
from huggingface_hub import snapshot_download

# 2. 下载到指定位置
model_path = snapshot_download(
    repo_id="distilbert/distilbert-base-uncased-finetuned-sst-2-english", 
    local_dir="./model_cache/distilbert-base-uncased-finetuned-sst-2-english",
    local_dir_use_symlinks=False  # 不使用符号链接，直接复制文件
)
print(model_path)
```

## 1.7 加载本地模型

```python
# 下载模型到本地
from huggingface_hub import snapshot_download

local_model_path = "./model_cache/finbert"
model_path = snapshot_download(
    repo_id="ProsusAI/finbert",
    local_dir=local_model_path,
    local_dir_use_symlinks=False # 不使用符号链接，直接复制文件
)
print(f"模型已下载到: {model_path}")
```

```python
# 加载模型
from transformers import pipeline
import os

# 使用本地模型
local_model_path = "./model_cache/finbert"

if os.path.exists(local_model_path):
    classifier = pipeline(
        'sentiment-analysis',
        model=local_model_path,
        tokenizer=local_model_path,
        # return_all_scores=True,
        top_k=None,
    )
    
    result = classifier("I love using Hugging Face libraries!")
    print(result)
else:
    print("本地模型不存在，请先下载模型")
```
